```python
#DETECTING OUTLIER for carat USING IQR METHOD
Q1=np.percentile(df['age'],25,method='midpoint')
Q2=np.percentile(df['age'],50,method='midpoint')
Q3=np.percentile(df['age'],75,method='midpoint')
print(Q1)
print(Q2)
print(Q3)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
#Finding upper_limit and lower_limit
upper_limit=Q3+1.5*(IQR)
lower_limit=Q1-1.5*(IQR)
print(upper_limit)
print(lower_limit)
```

```python
#Detecting outlier
outlier=[]
for i in df['age']:
    if ((i>upper_limit) or (i<lower_limit)):
        outlier.append(i)
outlier
```

```python
df1=(df['age']>upper_limit )| (df['age']<lower_limit)
df1
```

```python
x=df.loc[df1]
x
```
